```python
import seaborn as sns
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
#  Configuración
# ------------------------------------------------------------

model_names = {
    "BE": "Observada",
    "LC": "Lee-Carter",
    "DT": "Decision Tree",
    "RF": "Random Forest",
    "GB": "Gradient Boosting"
}

gender_names = {
    "Female": "Mujeres",
    "Male": "Hombres",
}

# Estilos de líneas (tuplas dashes)
model_styles = {
    "BE": "--",   # sólido
    "LC": "-",  # guiones
    "DT": ":",   # sólido
    "RF": ":",   # sólido
    "GB": ":"    # sólido
}

# Colores
model_colors = {
    "BE": "black",
    "LC": "grey",
    "DT": "green",
    "RF": "#0077FF",
    "GB": "red"
}

# Comparaciones
comparisons = [
    ["BE", "LC", "DT"],
    ["BE", "LC", "RF"],
    ["BE", "LC", "GB"]
]

# Carpeta
output_folder = os.path.join(os.path.expanduser("~"), "Downloads\Mort_Projs")

# ------------------------------------------------------------
#  Función principal
# ------------------------------------------------------------
def plot_3x2_comparisons(data, gender, year, variable="log_mx"):

    var_label = "ln(mx)" if variable == "log_mx" else "mx"

    fig, axes = plt.subplots(3, 2, figsize=(8, 10), sharey=False)
    fig.suptitle(f"{var_label} — {gender_names[gender]} — Año {year}", fontsize=14, y=1.02)

    age_ranges = [(0, 50), (50, 100)]

    for row_idx, modelos in enumerate(comparisons):

        for col_idx, (age_min, age_max) in enumerate(age_ranges):

            ax = axes[row_idx, col_idx]

            for m in modelos:

                df_plot = data[
                    (data["Model"] == m) &
                    (data["Year"] == year) &
                    (data["Gender"] == gender) &
                    (data["Age"].between(age_min, age_max))
                ].sort_values("Age")

                if df_plot.empty:
                    continue

                ax.plot(
                    df_plot["Age"],
                    df_plot[variable],
                    label=model_names[m],
                    color=model_colors[m],
                    linestyle=model_styles[m],
                    linewidth=1.5
                )

            ax.set_title(f"Edad {age_min}–{age_max}")
            ax.set_xlabel("Edad")
            ax.set_ylabel(var_label)
            ax.set_xticks(range(age_min, age_max+1, 5))
            ax.grid(True, alpha=0.25)
            ax.legend(loc="upper center", fontsize=8)

    plt.tight_layout()
    plt.savefig(f"{output_folder}/{var_label}_{gender}_{year}.png", dpi=300, bbox_inches="tight")
    plt.close()

    print(f"Imagen guardada: {output_folder}/{var_label}_{gender}_{year}.png")


# ------------------------------------------------------------
#  Loop final
# ------------------------------------------------------------
for gender in ["Male", "Female"]:
    for year in range(2011, 2020):
        plot_3x2_comparisons(df, gender, year, variable="log_mx")
        plot_3x2_comparisons(df, gender, year, variable="mx")
```
